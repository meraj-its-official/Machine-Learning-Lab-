# *Opening CSV file from URL*

```python
from io import StringIO
import pandas as pd
import requests

# Added 'raw.' before 'githubusercontent.com'
url = 'https://raw.githubusercontent.com/anhquan0412/dataset/main/aug_train.csv'
headers = {'User-Agent': 'Mozilla/5.0'}

response = requests.get(url, headers=headers)
response.raise_for_status()  # Good practice to check for HTTP errors

data = StringIO(response.text)
df = pd.read_csv(data)
df.head()

df.head()
```

# *Convertors*

```python
pd.read_csv('IPL_Matches_Data_2008_2026.csv', nrows=20)
```

```python
def rename(name):
    name = str(name).strip()  # Extra spaces hatane ke liye
    if name == "Royal Challengers Bangalore":
        return "RCB" 
    elif name == "Delhi Daredevils":
        return "DD"
    elif name == "Chennai Super Kings":
        return "CSK"
    elif name == "Rajasthan Royals":
        return "RR"
    elif name == "Kolkata Knight Riders":
        return "KKR"
    elif name == "Deccan Chargers":
        return "DC"
    elif name == "Mumbai Indians":
        return "MI"
    elif name == "Kings XI Punjab":
        return "KXP"
    else:
        return name
```

```python
pd.read_csv('IPL_Matches_Data_2008_2026.csv', converters={'team1':rename, 'team2':rename, 'toss_winner':rename} ,nrows=20)
```
