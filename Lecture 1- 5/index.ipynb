# Lecture 3: Linear Algebra Part 2

```python
# L-3
# Part 2 — Linear equations aur matrix inverse

A = np.array([
    [2.0, 1.0],
    [1.0, -1.0]
])

b = np.array([5.0, 1.0])

# Recommended numerical approach
x = np.linalg.solve(A, b)

print("Solution:", x)
print("Check A @ x:", A @ x)
print("Determinant:", np.linalg.det(A))
```

```python
# L-3
# Part 3 — Eigenvalues aur eigenvectors: Deep theory

A = np.array([
    [2.0, 0.0],
    [0.0, 3.0]
])

eigenvalues, eigenvectors = np.linalg.eig(A)

print("Eigenvalues:", eigenvalues)
print("Eigenvectors:\n", eigenvectors)

for i in range(len(eigenvalues)):
    v = eigenvectors[:, i]
    lam = eigenvalues[i]
    print("A @ v:", A @ v)
    print("lambda * v:", lam * v)
```

```python
# L-3
# Part 4 — SVD: Singular Value Decomposition
import numpy as np

A = np.array([
    [3.0, 0.0],
    [0.0, 2.0],
    [0.0, 0.0]
])

U, s, Vt = np.linalg.svd(A, full_matrices=False)

Sigma = np.diag(s)
A_reconstructed = U @ Sigma @ Vt

print("U shape:", U.shape)
print("Singular values:", s)
print("Vt shape:", Vt.shape)
print("Reconstructed matrix:\n", A_reconstructed)
print("Reconstruction error:",
      np.linalg.norm(A - A_reconstructed))
```

```python
# L-3
# Part 5 — Practical graph: PCA data ko compress kaise karta hai?
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

rng = np.random.default_rng(42)

# Two correlated features
z = rng.normal(size=200)
noise = rng.normal(scale=0.5, size=200)

X = np.column_stack([
    3 * z + noise,
    2 * z + rng.normal(scale=0.5, size=200)
])

# PCA with two components
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

# Keep only the first principal component
pca_1d = PCA(n_components=1)
X_1d = pca_1d.fit_transform(X)
X_reconstructed = pca_1d.inverse_transform(X_1d)

print("Original shape:", X.shape)
print("Reduced shape:", X_1d.shape)
print("Explained variance ratio:",
      pca_1d.explained_variance_ratio_[0])

# Plot the original observations
plt.figure(figsize=(7, 5))
plt.scatter(X[:, 0], X[:, 1], alpha=0.6)
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Original correlated data")
plt.axis("equal")
plt.grid(True)
plt.show()

# Plot the reconstruction from one component
plt.figure(figsize=(7, 5))
plt.scatter(X[:, 0], X[:, 1], alpha=0.35,
            label="Original")
plt.scatter(X_reconstructed[:, 0],
            X_reconstructed[:, 1],
            alpha=0.8, label="1-component reconstruction")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("PCA: reconstruction using one component")
plt.axis("equal")
plt.legend()
plt.grid(True)
plt.show()
```

# Lecture 4: Probability & Statistics for Machine Learning

```python
# L-4
# 4. Bayes' Theorem

p_spam = 0.20
p_not_spam = 1 - p_spam
p_offer_given_spam = 0.80
p_offer_given_not_spam = 0.10


# Total probability of observing "offer"
p_offer = (
    p_offer_given_spam * p_spam
    + p_offer_given_not_spam * p_not_spam
)

# Bayes theorem
p_spam_given_offer = (
    p_offer_given_spam * p_spam
) / p_offer

print("P(Offer):", p_offer)
print("P(Spam | Offer):", p_spam_given_offer)
print("Percentage:", p_spam_given_offer * 100)
```
